## 1. Empresa e problema

### 1.1 Empresa

A Receita Federal do Brasil (RFB) é um órgão vinculado ao Ministério da Fazenda, responsável pela administração dos tributos federais e pelo controle aduaneiro. Seu papel é essencial para a arrecadação de receitas da União, o combate à sonegação fiscal, à evasão de divisas e à lavagem de dinheiro. 

A RFB é uma das maiores instituições públicas do país, com abrangência nacional e uma base de dados extremamente ampla sobre contribuintes pessoas físicas e jurídicas.

Os “clientes” da Receita Federal são, de forma ampla, todos os cidadãos e empresas brasileiras que contribuem com tributos federais.

### 1.2 Problema

Apesar do aumento do acesso a dados econômicos no Brasil, a compreensão da desigualdade de renda entre regiões e ao longo do tempo ainda é limitada.

Mesmo com políticas públicas voltadas à redistribuição de renda (como o Bolsa Família, reajustes do salário mínimo e programas de crédito), a concentração de renda permanece elevada e varia significativamente entre os estados brasileiros.

**Problema central:** "Como os estados brasileiros se agrupam em relação aos perfis de renda dos contribuintes e características associadas (bens, despesas e impostos)?"

### 1.3 Como os dados podem ajudar a resolver o problema

A base de Distribuição de Renda da Receita Federal contém informações agregadas por centis da população declarantes (do 1º ao 100º centil), com indicadores como:
- Renda média;
- Renda total declarada;
- Quantidade de declarantes;
- Unidade da federação;
- Ano-base da declaração.

> Um centil é uma divisão de uma distribuição de dados em 100 partes iguais. Ou seja: se você ordenar todas as rendas das pessoas do menor para o maior valor, cada centil representa 1% da população, com base nessa ordem.

Com esses dados, é possível:
- Detectar perfis econômicos regionais;
- Comparar unidades federativas com características semelhantes;
- Visualizar padrões geográficos da renda declarada;
- Comunicar resultados de forma intuitiva a gestores e formuladores de políticas, orientando decisões baseadas em evidências.

Assim, os dados da Receita Federal permitem transformar informações fiscais agregadas em insights sobre o perfil socioeconômico dos estados brasileiros, apoiando diagnósticos e ações voltadas à redução de desigualdades regionais.

### 1.4 Questões éticas envolvidas

Embora a base seja pública e anonimizada, existem algumas questões éticas importantes a considerar:
- **Privacidade dos contribuintes:** Os dados são agregados por centis e não permitem identificar indivíduos, o que mitiga riscos de exposição. Ainda assim, é fundamental respeitar os limites de granularidade e não tentar inferir características pessoais ou socioeconômicas individuais.
- **Uso responsável das informações:** A interpretação incorreta dos dados (por exemplo, tirar conclusões causais indevidas) pode levar a distorções na comunicação pública. Deve-se deixar claro que os dados representam apenas pessoas físicas declarantes de IR, não toda a população brasileira.
- **Viés de representatividade:** Como apenas uma parte da população brasileira declara imposto de renda (geralmente as faixas mais altas de renda), as conclusões não representam todo o espectro social.

## 5. Realizar a modelagem

Será realizada uma análise não supervisionada com clustering K-means para agrupar estados semelhantes em termos de distribuição de renda. 

Essa análise é útil para identificar padrões e segmentar regiões com características econômicas similares, o que pode auxiliar na formulação de políticas públicas direcionadas.


### 5.1 Seleção das variáveis

````python
cols = [
    'Rendimentos Tributaveis - Soma da RTB do Centil [R$ milhões]',
    'Rendimentos Tributaveis - Média da RTB do Centil [R$]',
    'Rendimentos Sujeitos à Tribut. Exclusiva [R$ milhões]',
    'Rendimentos Isentos - Lucros e dividendos [R$ milhões]',
    'Rendimentos Isentos - Rendim. Sócio/Titular ME/EPP Opt SIMPLES [R$ milhões]',
    'Rendimentos Isentos - Outros Rendimentos Isentos [R$ milhões]',
    'Despesas Dedutíveis - Previdência [R$ milhões]',
    'Despesas Dedutíveis - Dependentes [R$ milhões]',
    'Despesas Dedutíveis - Instrução [R$ milhões]',
    'Despesas Dedutíveis - Médicas [R$ milhões]',
    'Despesas Dedutíveis - Pensão Alimentícia [R$ milhões]',
    'Despesas Dedutíveis - Livro-Caixa [R$ milhões]',
    'Imposto Devido [R$ milhões]',
    'Bens e Direitos - Imóveis [R$ milhões]',
    'Bens e Direitos - Móveis [R$ milhões]',
    'Bens e Direitos - Financeiros [R$ milhões]',
    'Bens e Direitos - Outros Bens e Direitos [R$ milhões]',
    'Dívidas e Ônus [R$ milhões]'
]

df_grouped = df.groupby('Ente Federativo')[cols].mean()

df_grouped.head()
````

### 5.2 Padronização dos dados

````python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_grouped)
````

### 5.3 Método do cotovelo

````python
inertia = []
K = range(1, 11)

for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(K, inertia, 'bo-')
plt.xlabel('Número de Clusters (k)')
plt.ylabel('Inércia')
plt.title('Método do Cotovelo para Determinar o k Ótimo')
plt.show()
````

### 5.4 Aplicação do K-means

````python
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df_grouped['Cluster'] = kmeans.fit_predict(X_scaled)
````

### 5.5 Análise dos resultados

````python
# Perfis por cluster (médias das variáveis selecionadas)
cluster_profile = df_grouped.groupby('Cluster')[cols].mean()
print("Perfis médios por cluster (médias das variáveis selecionadas):")
display(cluster_profile)

# Centróides padronizados (importante para ver quais features importam)
centroids_std = pd.DataFrame(kmeans.cluster_centers_, columns=df_grouped.columns[:-1])  # sem a coluna 'Cluster'
print("Centroides (dados padronizados):")
display(centroids_std.T.sort_values(by=0, ascending=False))  # exemplo de ordenação
# ...existing code...
# Nomeação qualitativa dos clusters com base nos centróides padronizados
centroids = centroids_std.copy()  # centróides já padronizados (z-score-like)

def assign_label(centroid):
    # regras simples por palavras-chave; ajustar thresholds conforme necessário
    if centroid.filter(like='Média').max() > 0.8 and centroid.filter(like='Bens e Direitos').max() > 0.5:
        return 'Alta renda / Alto patrimônio'
    if centroid.filter(like='Média').max() < -0.8:
        return 'Baixa renda'
    if centroid.filter(like='Dívidas').max() > 0.8:
        return 'Endividamento elevado'
    if centroid.filter(like='Imposto Devido').max() > 0.8:
        return 'Alta carga tributária'
    return 'Perfil intermediário / misto'

labels = {}
print("Resumo qualitativo por cluster:\n")
for i in centroids.index:
    s = centroids.loc[i]
    top_pos = s.sort_values(ascending=False).head(3)
    top_neg = s.sort_values().head(3)
    label = assign_label(s)
    labels[i] = label
    print(f"Cluster {i} -> {label}")
    print("  Top + :", ', '.join([f"{f} ({v:.2f})" for f,v in top_pos.items()]))
    print("  Top - :", ', '.join([f"{f} ({v:.2f})" for f,v in top_neg.items()]))
    print()

# opcional: adicionar rótulo ao df_grouped
df_grouped['Cluster_label'] = df_grouped['Cluster'].map(labels)
display(df_grouped[['Cluster','Cluster_label']])
# ...existing code...

# # Silhouette score (avaliação da separação)
# from sklearn.metrics import silhouette_score
# sil_score = silhouette_score(X_scaled, df_grouped['Cluster'])
# print(f"Silhouette score: {sil_score:.3f}")

# # PCA 2D para visualização dos Ente Federativo no espaço reduzido
# from sklearn.decomposition import PCA
# pca = PCA(n_components=2, random_state=42)
# proj = pca.fit_transform(X_scaled)
# df_plot = df_grouped.reset_index().copy()
# df_plot['PC1'] = proj[:,0]
# df_plot['PC2'] = proj[:,1]

# plt.figure(figsize=(8,6))
# sns.scatterplot(data=df_plot, x='PC1', y='PC2', hue='Cluster', palette='tab10', s=100)
# for i, r in df_plot.iterrows():
#     plt.text(r['PC1']+0.02, r['PC2']+0.02, r['Ente Federativo'], fontsize=8)
# plt.title('PCA 2D - Estados por Cluster')
# plt.show()

# # Tabela de concentração: para cada estado, proporção de presença por cluster
# # (usar df original para contar centis por estado; caso o agrupamento original seja por estado médio, adaptar)
# state_cluster_counts = df.groupby(['Ente Federativo']).apply(lambda g: g.index)  # placeholder se df tem linhas por centil
# # mais direto: se quiser a partir do df original com centis e clusters por linha, compute df['Cluster_row'] antes
# # aqui, se só tiver clusters por estado (df_grouped), mostramos distribuição 100% para o cluster atribuído
# state_cluster_dist = df_grouped.reset_index()[['Ente Federativo','Cluster']].copy()
# state_cluster_dist['Share'] = 1.0  # cada estado está inteiramente no seu cluster (quando clustering foi em médias)
# print("Distribuição (por estado) — quando clustering foi feito sobre médias por estado cada estado pertence a 1 cluster:")
# display(state_cluster_dist.sort_values(['Cluster','Ente Federativo']))
````
